```python
import numpy as np
import matplotlib.pyplot as plt
import random

# Data points
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T

# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")

plt.scatter(X, Y)

Y = m*X + c

plt.plot(X, Y, 'r')
plt.show()
```

```python
# Data points
X = np.sort(np.random.rand(5))
Y = np.sort(np.random.rand(5))
# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T
# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")


plt.scatter(X, Y)

Y = m*X + c

plt.plot(X, Y, 'r')
plt.show()
```

```python
#With the vectors randomized, the code still creates an accurate line of best fit.
```

```python
# Data points
X = np.sort(np.random.rand(5))
Y = np.sort(np.random.rand(5))
# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T
# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")

plt.subplot(1, 2, 1)
plt.scatter(X, Y)
Y = m*X + c
plt.plot(X, Y, 'r')


X = np.arange(0, 1.01, 0.01)
Y = m*X + c + random.uniform(-1, 1)
A = np.vstack([X, np.ones(len(X))]).T
# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope(New): {m}, Intercept(New): {c}")

plt.subplot(1, 2, 2)
plt.plot(X, Y, 'g')
plt.show()
```
