```python
import math
import random
import numpy
import matplotlib.pyplot as plt

def montecarlo_pi(number_of_shots):
    number_of_hits = 0
    for i in range(0,number_of_shots):
        x=random.uniform(-1,1)
        y=random.uniform(-1,1)
        if (x*x + y*y <1):
            number_of_hits = number_of_hits + 1
        
    return (4*number_of_hits/number_of_shots)

#Monte Carlo Pi function
```

```python
McpPlot = []


for i in range(10, 1001, 20):
    McpError = abs((montecarlo_pi(i)-numpy.pi))
    McpPlot.append(round(McpError, 15))
    
#Plotting of the calculated error from 10 to 1000 (1001 is listed in the range function because it is exclusive)

plt.plot(McpPlot, 'g')
plt.show()
```
