```python
import math
import random
import numpy
import matplotlib.pyplot as plt


#Defining the generalization of the Taylor Series for sin(x)

def taylor_series(value, numTerm):
    calculated_sin = 0
    n = 0
    for n in range (numTerm):
        calculated_sin += ((-1)**n)*(value**(2*n+1))/(math.factorial(2*n+1))
    return ()


oneTerm = []
twoTerm = []
threeTerm = []
fourTerm = []
realSin = []
absError1 = []
absError2 = []
absError3 = []
absError4 = []


for i in range (0, 10):
    oneTerm.append(taylor_series(5, 1))
    print(oneTerm)
    twoTerm.append(taylor_series(i, 2))
    threeTerm.append(taylor_series(i, 3))
    fourTerm.append(taylor_series(i, 4))
    realSin.append(numpy.sin(i))
    absError1.append(taylor_series(i, 1) - numpy.sin(i))
    absError2.append(taylor_series(i, 2) - numpy.sin(i))
    absError3.append(taylor_series(i, 3) - numpy.sin(i))
    absError4.append(taylor_series(i, 4) - numpy.sin(i))
  
    
#plt.plot(oneTerm, 'g', absError1, 'r')
#plt.show()
```
